```python
# 환경 확인과 4개 CSV 불러오기
# customers
# orders

from course_utils.paths import get_data_dir
import pandas as pd

customers = pd.read_csv(get_data_dir()/"raw/customers.csv")
orders = pd.read_csv(get_data_dir()/"raw/orders.csv")

print(customers.info())
print(orders.info())
```

```python
# products
# order_items

products = pd.read_csv(get_data_dir()/"raw/products.csv")
order_items = pd.read_csv(get_data_dir()/"raw/order_items.csv")

print(products.info())
print(order_items.info())
```

```python
# payment_method
# 주문상태의 종류는 몇가지이고 각 상태별 주문 수량?
# 결재수단의 종류는 몇가지이고 각 결재 상태별 주문 수량?

print(orders.value_counts("order_status"))
print(orders.value_counts("payment_method"))

ret = orders["order_status"].value_counts(dropna=False)
print(type(ret))
print(ret)
```

```python
# customers_over_30

# customers_over_30 = customers[customers.age >= 30]
customers_over_30 = customers[customers["age"] >= 30]

print(customers_over_30)
```

```python
# 나이가 제일 많은 사람
print(customers_over_30["age"].max())

# 나이가 제일 적은 사람
print(customers_over_30["age"].min())

# 나이에 대한 통계 정보
print(customers_over_30["age"].describe())

# customers_over_30 의 통계 정보
print(customers_over_30.describe())
```

```python
# 도시별 회원 수

print(customers["city"].value_counts(dropna=False))
```

```python
# 서울, 부산, 인천 도시의 고객들은 누구인가?

city_customers = customers[customers["city"].isin(["서울", "부산", "인천"])]

print(city_customers.head(10))
print(city_customers["city"].value_counts())
```

```python
# 컬럼 추가(파생 컬럼)
# 라인 합계 금액 : line_total = quantity * unit_price

order_items["line_total"] = order_items["quantity"] * order_items["unit_price"]

print(order_items.head())

print(order_items.info())
```

```python
# 중복 여부 확인

print(orders["order_id"].duplicated().sum())
```

```python
# order_items
# order_status를 함께 출력하려면 => merge

order_sales = order_items.merge(
    orders[["order_id", "customer_id", "order_date", "order_status"]],
    on="order_id",
    how="left",
    validate="many_to_one"
)

print(order_sales.head())
print(order_sales.info())
```

```python
order_sales_test = order_sales.copy()

print(order_sales_test.info())
```

```python
order_sales_test["order_date"] = pd.to_datetime(
    order_sales_test["order_date"], errors="coerce",
)

print(order_sales_test.info())
```
