```python
# 현재 폴더를 확인한다.
PROJECT_ROOT = Path.cwd()
# 현재 폴더 안에 data 폴더가 있는지 확인한다.
while True:
    print(f"PROJECT_ROOT: {PROJECT_ROOT}")
    # 만약 data 폴더가 있다면
    if (PROJECT_ROOT / "data").is_dir():
    #    현재 폴더를 프로젝트 루트로 결정한다.
        print(f"This is the PROJECT_ROOT: {PROJECT_ROOT}")
    #    탐색을 종료한다.
        break
    # 만약 data 폴더가 없다면
    PROJECT_ROOT = PROJECT_ROOT.parent
    #    부모 폴더로 한 단계 올라간다.


# 다시 data 폴더가 있는지 확인한다.
# 이 과정을 반복한다.
# 더 이상 위로 올라갈 수 없는데도
# 데이터 폴더를 찾지 못했다면
# 오류를 발생시킨다.
```

```python
from pathlib import Path

# 현재 폴더를 확인한다.
def get_project_root():
    PROJECT_ROOT = Path.cwd()
    while True:
        if (PROJECT_ROOT / "data").is_dir():
            break

        PROJECT_ROOT = PROJECT_ROOT.parent

    return PROJECT_ROOT

def get_data_dir():
    return get_project_root() / "data"

print(get_data_dir())
```
